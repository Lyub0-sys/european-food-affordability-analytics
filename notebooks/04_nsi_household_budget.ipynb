```python
connection = get_connection()
cursor = connection.cursor()

cursor.execute("""
    SELECT
        year,
        food_expenditure_share_pct
    FROM household_food_expenditure_share
    WHERE geo = 'BG'
    ORDER BY year;
""")

rows = cursor.fetchall()

nsi_db_df = pd.DataFrame(
    rows,
    columns=["year", "food_expenditure_share_pct"]
)

cursor.close()
connection.close()

display(nsi_db_df)
```

```python
import matplotlib.pyplot as plt

nsi_db_df["food_expenditure_share_pct"] = (
    nsi_db_df["food_expenditure_share_pct"].astype(float)
)

plt.plot(
    nsi_db_df["year"],
    nsi_db_df["food_expenditure_share_pct"],
    marker="o"
)

plt.title("Food and non-alcoholic beverages share of household monetary expenditure")
plt.xlabel("Year")
plt.ylabel("Share of monetary expenditure (%)")

plt.xticks(
    nsi_db_df["year"],
    rotation=45
)

plt.grid(True)
plt.show()
```

## Interpretation and methodological note

The share of food and non-alcoholic beverages in Bulgarian household monetary expenditure decreased from 34.9% in 2008 to 29.1% in 2025.

The overall trend is downward, although it is not continuous. The series reached 35.8% in 2010, declined substantially during the following years, and reached its lowest observed value of 28.3% in 2024 before increasing to 29.1% in 2025.

A lower food expenditure share may indicate a reduced relative food burden within the household budget. However, this indicator should not be interpreted as a direct measure of food affordability or household welfare.

The expenditure share depends on both food expenditure and total household monetary expenditure. Changes may therefore reflect movements in food prices, household incomes, consumption patterns, or other expenditure categories.

In this project, the NSI household expenditure share is used as a complementary indicator to the Eurostat-based affordability change proxy.

- Eurostat HICP and income data describe changes in food prices and income dynamics.
- NSI household budget data describe the observed share of household monetary expenditure allocated to food and non-alcoholic beverages.
- Neither indicator alone establishes causality.
