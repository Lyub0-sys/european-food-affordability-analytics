```python
import requests

url = "https://ec.europa.eu/eurostat/api/dissemination/statistics/1.0/data/ilc_di03"

params = {}

response = requests.get(url, params=params, timeout=10)
response.raise_for_status()

data = response.json()

print(type(data))
```
